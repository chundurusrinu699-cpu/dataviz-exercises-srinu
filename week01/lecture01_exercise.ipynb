```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio
```

```python
df_stocks = px.data.stocks()
df_stocks.head()
```

```python
df_stocks.info()
```

```python
fig = px.line(
    df_stocks,
    x='date',
    y=['AAPL', 'AMZN'],
    title='Apple vs Amazon Stock Performance'
)

fig.show()
```

```python
df_tidy = df_stocks.melt(
    id_vars='date',
    var_name='company',
    value_name='value'
)

df_tidy.head()
```

```python
fig = px.line(
    df_tidy,
    x='date',
    y='value',
    color='company',
    title='Stock Performance Comparison'
)

fig.show()
```

```python
pio.templates.default = "plotly_white"

fig = px.line(
    df_tidy,
    x='date',
    y='value',
    color='company',
    title='Technology Stocks Performance'
)

fig.update_layout(
    xaxis_title='Date',
    yaxis_title='Relative Stock Value',
    legend_title='Company'
)

fig.show()
```
